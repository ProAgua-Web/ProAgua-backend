### Código de remoção de linhas
A seguir, temos um trecho de código para remoção das linhas citadas anteriormente.

```python
# Criar uma cópia da base de dados
clean_df = df.copy()

# Remover as 5 primeiras linhas (cabeçalho)
clean_df = df.iloc[5:]  

# Conjunto de valores indesejados no dataset
tipos_invalidos = [
    "Não houve coleta (Prédio fechado)",
    "Não  houve coleta (Prédio fechado)",
    "Não houve coleta (Prédio sem água)",
    "Não houve coleta em função de mal entendido pelo TCC de Camila",
    "Não houve coleta (Banheiro feminino sem água)",
    "Não houve coleta ",
]

# Remover linhas que possuem os valores contidos no conjunto
clean_df = clean_df[~clean_df.iloc[:, 6].isin(tipos_invalidos)]
```

Em algumas sequências de coletas não há dados de pontos ou coletas. Todas as sequências que não possuem dados do ponto de coleta, também não possuem coletas.

```python
# Verificar quais linhas possuem dados faltantes do ponto de coleta incial
contains_null = clean_df.iloc[:, 6:8].isnull().aggregate(lambda x: x.any(), axis=1)
clean_df = clean_df.loc[~contains_null]
```

### Padronização

```python
def data_processing(data: str):
    if type(data) == str:
        return data.lower().lstrip().rstrip()
    if data is float('nan'):
        return None
```

```python
from typing import Any

def mapper(map: dict[Any, Any]):
    def func(value):
        if value in map:
            return map[value]
        return value
    return func
```

```python
map = {
    "a": "ausência"
}

clean_df.loc[:, '"Escherichia coli" (presença/ ausência)'] = clean_df.loc[:, '"Escherichia coli" (presença/ ausência)'].map(data_processing).map(mapper(map)) # type: ignore
set(clean_df.loc[:, '"Escherichia coli" (presença/ ausência)'].to_numpy().flatten().tolist())
```

```python
clean_df.loc[:, 'Coliformes totais (presença/ ausência)'] = clean_df.loc[:, 'Coliformes totais (presença/ ausência)'].map(data_processing).map(mapper(map)) # type: ignore
set(clean_df.loc[:, 'Coliformes totais (presença/ ausência)'].to_numpy().flatten().tolist())
```

```python
clean_df.loc[:, "Amostragem"] = clean_df.loc[:, "Amostragem"].map(data_processing) # type: ignore
clean_df.loc[:, "Amostragem"].unique()
```

```python
clean_df.loc[:, "Campus(Leste ou Oeste)"] = clean_df.loc[:, "Campus(Leste ou Oeste)"].map(data_processing) # type: ignore
clean_df.loc[:, "Campus(Leste ou Oeste)"].unique()
```

```python
clean_df.loc[:, "Tipo"] = clean_df.loc[:, "Tipo"].map(data_processing) # type: ignore
clean_df.loc[:, "Tipo"].unique()
```

```python
map = {
    "tormeira (banheiro masculino)": "torneira"
}
clean_df.loc[:, "Tipo"] = clean_df.loc[:, "Tipo"].map(mapper(map)) # type: ignore
clean_df.loc[:, "Tipo"].unique()
```

### Padronizar datas

```python
import datetime

def to_date(value):
    fmt ="%d/%m/%Y"
    
    if type(value) == str:
        value = datetime.datetime.strptime(value, fmt).date()
    elif type(value) == datetime.datetime:
        value = value.date()
    else:
        return None

    return value
```

```python
# Padronizar tipo de todas as colunas com nome "Data"
clean_df.loc[:, "Data"] = clean_df.loc[:, "Data"].map(to_date) # type: ignore
clean_df.loc[:, "Data"]
```

### Padronizar valores float

```python
clean_df.loc[:, "Temperatura (˚C)"].to_numpy()
```

```python
import re
import warnings

def to_float(value: str | float | int):
    if type(value) == str:
        pattern = r'\d+[,.]\d+'
        if re.match(pattern, value):
            value = value.replace(',', '.')
            return float(value)
        else:
            warnings.warn(f"The value \"{value}\" didnt match the float pattern")

    elif type(value) == float or type(value) == int:
        return float(value)
    else:
        warnings.warn(f"Inválid type {type(value)} {value}")

    return float('nan')
```

```python
clean_df.loc[:, 'Temperatura (˚C)'] = clean_df.loc[:, "Temperatura (˚C)"].map(to_float)
clean_df.loc[:, 'Cloro residual livre (mg/L)'] = clean_df.loc[:, 'Cloro residual livre (mg/L)'].map(to_float)
clean_df.loc[:, 'Turbidez (uT)'] = clean_df.loc[:, 'Turbidez (uT)'].map(to_float)
clean_df.loc[:, 'Cor '] = clean_df.loc[:, 'Cor '].map(to_float)
```
